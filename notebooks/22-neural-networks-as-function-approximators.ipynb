# Blog 22 — Can a Neural Network Learn Any Shape?

```python
import torch
m=torch.nn.Sequential(torch.nn.Linear(1,16),torch.nn.Tanh(),torch.nn.Linear(16,1))
x=torch.linspace(-2,2,100).unsqueeze(1); y=x**3
opt=torch.optim.Adam(m.parameters(),lr=.02)
for _ in range(500): opt.zero_grad(); loss=((m(x)-y)**2).mean(); loss.backward(); opt.step()
print('final loss:',loss.item())
```
