## 3. Connect the picture to the math

For many deep-learning ideas, the same pattern appears:

**numbers → transformation → prediction → error → improvement**

The exact transformation changes from blog to blog, but the learning loop stays recognizable.

### Experiment: change one thing
Run the next cell, then change one number at a time. Watch the graph and explain what changed. This is the scientific loop used throughout machine learning: **predict → measure → change → measure again**.


````python
def score(parameter):
    # A deliberately simple objective: its minimum is easy to see.
    return (parameter - 1.5) ** 2 + 0.2

parameters = np.linspace(-3, 5, 200)
losses = score(parameters)

plt.plot(parameters, losses)
plt.scatter([1.5], [0.2], s=70, label="best point")
plt.xlabel("parameter")
plt.ylabel("score / loss")
plt.legend()
plt.title("A landscape we can explore")
plt.show()
````

### Visual intuition
The animation below makes an invisible learning process visible. It is intentionally small so it can run in Colab without special hardware.

```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

xs = np.linspace(-3, 3, 120)
frames = np.linspace(-2.5, 2.5, 60)
fig, ax = plt.subplots()
ax.set_xlim(-3, 3); ax.set_ylim(-1, 10)
line, = ax.plot([], [])

def draw(i):
    c = frames[i]
    y = (xs - c) ** 2
    line.set_data(xs, y)
    ax.set_title(f'Learning step {i + 1}: parameter = {c:.2f}')
    return line,

ani = FuncAnimation(fig, draw, frames=len(frames), interval=60, blit=True)
HTML(ani.to_jshtml())
```
