## Learning objectives
Represent examples as vectors, perform vector arithmetic, calculate norms and distances, compute dot products, visualize vectors, and investigate feature scaling.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. A vector is an ordered collection
Alice has maths and science scores `[9, 8]`. The position of each number matters.

```python
alice = np.array([9.0, 8.0])
bob = np.array([8.0, 9.0])
cara = np.array([3.0, 4.0])
print('Alice:', alice)
print('shape:', alice.shape)
assert alice.shape == (2,)
```

## 2. Vector arithmetic
Hand calculation: `[2,1] + [1,3] = [3,4]` and `2[2,1] = [4,2]`. Now let NumPy verify it.

```python
a = np.array([2.0, 1.0])
b = np.array([1.0, 3.0])
print('a + b =', a + b)
print('2a =', 2 * a)
assert np.allclose(a + b, [3, 4])
assert np.allclose(2 * a, [4, 2])
```

## 3. Norm and distance
The Euclidean norm is `sqrt(x1² + x2² + ...)`. Distance is the norm of the difference.

```python
x = np.array([3.0, 4.0])
y = np.array([0.0, 0.0])
manual_norm = np.sqrt(np.sum(x ** 2))
numpy_norm = np.linalg.norm(x)
distance = np.linalg.norm(x - y)
print('manual norm:', manual_norm)
print('NumPy norm:', numpy_norm)
print('distance:', distance)
assert np.isclose(manual_norm, 5.0)
assert np.isclose(numpy_norm, 5.0)
```

## 4. Dot product
For `x=[2,3]` and `w=[4,5]`: `2×4 + 3×5 = 23`. This weighted sum will become the core calculation inside a neuron.

```python
x = np.array([2.0, 3.0])
w = np.array([4.0, 5.0])
manual_dot = x[0] * w[0] + x[1] * w[1]
numpy_dot = x @ w
print('manual:', manual_dot)
print('NumPy:', numpy_dot)
assert np.isclose(manual_dot, 23.0)
assert np.isclose(numpy_dot, 23.0)
```

## 5. Visualize vectors
The same vector is both numbers and an arrow. Change the coordinates and rerun the cell.

```python
vectors = [(np.array([2.0, 1.0]), 'a'), (np.array([1.0, 3.0]), 'b'), (np.array([3.0, 4.0]), 'a+b')]
fig, ax = plt.subplots(figsize=(6, 6))
for v, label in vectors:
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1)
    ax.text(v[0], v[1], label)
ax.set_xlim(-1, 5); ax.set_ylim(-1, 5); ax.set_aspect('equal'); ax.grid(True)
ax.set_xlabel('x'); ax.set_ylabel('y'); ax.set_title('Vectors as arrows')
plt.show()
```

## 6. Dot product as geometry
`x·w = ||x||||w||cos(theta)`. Zero means perpendicular; positive means an acute angle; negative means an obtuse angle.

```python
u = np.array([1.0, 0.0])
v = np.array([1.0, 1.0])
cos_theta = (u @ v) / (np.linalg.norm(u) * np.linalg.norm(v))
theta = np.degrees(np.arccos(np.clip(cos_theta, -1, 1)))
print('angle:', theta, 'degrees')
assert np.isclose(theta, 45.0)
```

## 7. Experiment — feature scaling changes geometry
Change one feature by a factor of 100. The underlying objects did not change, but the geometry used by a distance calculation did.

```python
p = np.array([10.0, 2.0]); q = np.array([12.0, 3.0])
before = np.linalg.norm(p - q)
scale = np.array([1.0, 100.0])
after = np.linalg.norm(p * scale - q * scale)
print('before:', before)
print('after:', after)
assert after > before
```

**Scientist's question:** Did the objects change, or did our coordinate system change? What does this tell you about preprocessing and representation?

## 8. Many vectors form a matrix

```python
X = np.array([[2, 3, 4], [5, 6, 7], [8, 9, 10]], dtype=float)
print(X)
print('shape:', X.shape)
assert X.shape == (3, 3)
```

## 9. Optional PyTorch bridge
The mathematics is unchanged; PyTorch provides tensors and will later provide automatic differentiation.

```python
try:
    import torch
    tx = torch.tensor([2.0, 3.0]); tw = torch.tensor([4.0, 5.0])
    value = torch.dot(tx, tw).item()
    print('PyTorch dot:', value)
    assert np.isclose(value, 23.0)
except ImportError:
    print('PyTorch is optional for this lesson.')
```
