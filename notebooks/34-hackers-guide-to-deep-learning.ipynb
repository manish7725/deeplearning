# Blog 34 — The Detective's Guide to Training a Neural Network

```python
import torch
torch.manual_seed(0); model=torch.nn.Linear(1,1); x=torch.tensor([[1.],[2.]]); y=torch.tensor([[2.],[4.]])
opt=torch.optim.SGD(model.parameters(),lr=.1)
for step in range(20):
    opt.zero_grad(); loss=((model(x)-y)**2).mean(); loss.backward(); opt.step()
print('loss:',loss.item())
```
