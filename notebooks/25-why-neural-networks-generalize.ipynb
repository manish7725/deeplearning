### Experiment: change one thing
Run the next cell, then change one number at a time. Watch the graph and explain what changed. This is the scientific loop used throughout machine learning: **predict → measure → change → measure again**.


### 🔎 Code tutorial — what to notice

- The first important objects are the data. Look at their shapes and values before thinking about the model. In deep learning, the shape tells you how many examples and features are being processed.

- The function packages a repeated calculation into a named step. Read its inputs first, then its return value; the function name is a clue about the mathematical operation.

- These variables are parameters: numbers the learning algorithm is allowed to change. Ask: what would happen if this number became slightly larger?

- This measures how far the prediction is from the target. A smaller value means the model is doing better according to this chosen objective.

- The plotting code is not decoration: it turns numbers into evidence. Before running it, predict the shape of the graph; after running it, explain why the graph has that shape.

````python
def score(parameter):
    # A deliberately simple objective: its minimum is easy to see.
    return (parameter - 1.5) ** 2 + 0.2

parameters = np.linspace(-3, 5, 200)
losses = score(parameters)

plt.plot(parameters, losses)
plt.scatter([1.5], [0.2], s=70, label="best point")
plt.xlabel("parameter")
plt.ylabel("score / loss")
plt.legend()
plt.title("A landscape we can explore")
plt.show()
````

from matplotlib.animation import FuncAnimation
from IPython.display import HTML

xs = np.linspace(-3, 3, 120)
frames = np.linspace(-2.5, 2.5, 60)
fig, ax = plt.subplots()
ax.set_xlim(-3, 3); ax.set_ylim(-1, 10)

def draw(i):

    return line,

ani = FuncAnimation(fig, draw, frames=len(frames), interval=60, blit=True)

## 2. 🧮 NumPy — build the idea yourself

NumPy is deliberately low-level here. Do not hide the mathematics behind a high-level library. Print the intermediate arrays and connect each one to a symbol in the blog.

````python
# NumPy practice scaffold — Chapter 25
# change model complexity and compare the gap
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
x = np.linspace(-3, 3, 101)
y = np.sin(x)

z = x**2
print("x shape:", x.shape)
print("first five x:", x[:5])
print("first five z:", z[:5])

plt.figure(figsize=(8,4))
plt.plot(x, y, label="reference")
plt.plot(x, z / z.max(), label="normalized experiment")
plt.title("Chapter 25: predict, run, explain")
plt.legend(); plt.show()
````

## 3. 🔥 PyTorch — express the same mathematics with tensors

Now repeat the calculation with PyTorch. The point is not to replace mathematics with a framework. The point is to prove that the framework is executing the mathematics you already understand.

````python
import torch

torch.manual_seed(7)
tx = torch.linspace(-3, 3, 101)
tx.requires_grad_(True)
tz = tx**2
print("tensor shape:", tuple(tx.shape))
print("first five values:", tz[:5])

probe = tz.mean()
probe.backward()
print("gradient exists:", tx.grad is not None)
print("first five gradients:", tx.grad[:5])
````

## 4. 📈 Visualize what the mathematics predicts

A graph is an experiment, not decoration. Before running the next cell, predict: **where should the curve be high, low, steep, or flat?** Then compare your prediction with the plot.

````python
plt.figure(figsize=(8,4))
plt.plot(tx.detach().numpy(), tz.detach().numpy())
plt.xlabel("input")
plt.ylabel("mathematical output")
plt.title("What the equation predicts")
plt.grid(True)
plt.show()
````
