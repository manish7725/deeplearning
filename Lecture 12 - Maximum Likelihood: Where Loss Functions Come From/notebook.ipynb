## Prediction
For 8 heads and 2 tails, predict the best value of `p` before running the code.

```python
import math

heads, tails = 8, 2
p = 0.8
likelihood = p**heads * (1-p)**tails
likelihood
```

## Mathematics

`L(p) = p^8 (1-p)^2` and `loss(p) = -log(L(p))`. The log is monotone, so both choose the same `p`. 

```python
p = 0.8
manual = p**8 * (1-p)**2
assert math.isclose(manual, 0.026843545600000006, rel_tol=1e-12)
print(manual)
```

```python
def likelihood(p, heads, tails):
    return p**heads * (1-p)**tails

assert likelihood(0.5, 8, 2) == 0.01953125
assert math.isclose(likelihood(0.8, 8, 2), manual)
```

```python
import matplotlib.pyplot as plt

ps = [i/100 for i in range(1,100)]
ls = [likelihood(p, heads, tails) for p in ps]
plt.plot(ps, ls)
plt.xlabel('p')
plt.ylabel('Likelihood')
plt.title('Likelihood for 8 heads and 2 tails')
plt.show()
```

```python
best_p = max(ps, key=lambda p: likelihood(p, heads, tails))
print('Grid-search best p:', best_p)
assert abs(best_p - 0.8) <= 0.01
```

```python
heads2, tails2 = 3, 7
best_p2 = max(ps, key=lambda p: likelihood(p, heads2, tails2))
print('New best p:', best_p2)
assert abs(best_p2 - 0.3) <= 0.01
```

## Observe
The peak moves from about `0.8` to `0.3` when the observed number of heads changes from 8 to 3.

## Explain
For Bernoulli data the log-likelihood is `H log p + T log(1-p)`. Setting its derivative to zero gives `p = H/(H+T)`. The frequency is therefore derived by maximum likelihood.
