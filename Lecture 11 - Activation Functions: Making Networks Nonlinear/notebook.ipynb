# Lecture 11 — Activation Functions: Making Networks Nonlinear


```python
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-5, 5, 400)
relu = np.maximum(0, z)
sigmoid = 1 / (1 + np.exp(-z))

plt.figure(figsize=(8,4))
plt.plot(z, relu, label='ReLU')
plt.plot(z, sigmoid, label='Sigmoid')
plt.axhline(0, linewidth=0.8)
plt.legend()
plt.title('Activation functions')
plt.xlabel('z')
plt.ylabel('activation')
plt.show()
```

## Compare values

```python
values = np.array([-2., -0.5, 0., 1., 3.])
print('z       :', values)
print('ReLU    :', np.maximum(0, values))
print('Sigmoid :', 1/(1+np.exp(-values)))
```

## Why two linear layers collapse

```python
W1 = np.array([[1., 2.], [0., 1.]])
W2 = np.array([[2., 1.], [1., 3.]])
x = np.array([3., 4.])
print('two linear layers:', W2 @ (W1 @ x))
print('single combined layer:', (W2 @ W1) @ x)
```

## Add ReLU and the collapse disappears in general

```python
def relu(x):
    return np.maximum(0, x)

print('linear -> ReLU -> linear:', W2 @ relu(W1 @ x))
```

## XOR experiment

A linear classifier cannot separate XOR with one straight boundary.

```python
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,1,1,0])
print('X:', X)
print('targets:', y)
print('A single line cannot separate these labels.')
```

## PyTorch activations

```python
import torch
import torch.nn as nn

x_t = torch.tensor([-2.0, -0.5, 1.0, 3.0])
print('ReLU:', nn.ReLU()(x_t))
print('Sigmoid:', nn.Sigmoid()(x_t))
print('Tanh:', nn.Tanh()(x_t))
```

## Challenge

For `L(x) = (ReLU(x) - 2)^2`, test x=-1 and x=1. Then reason about the gradient on each side of zero.

```python
for value in [-1.0, 1.0]:
    x = torch.tensor(value, requires_grad=True)
    loss = (torch.relu(x) - 2)**2
    loss.backward()
    print(f'x={value}, loss={loss.item()}, dL/dx={x.grad.item()}')
```
