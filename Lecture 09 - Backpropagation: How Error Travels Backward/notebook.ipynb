# Lecture 09 — Backpropagation: How Error Travels Backward

This notebook follows the blog and derives backpropagation from the chain rule.


```python
import math

# A tiny computation graph:
# z = 2x, y = z + 3, L = y^2
x = 1.0
z = 2*x
y = z + 3
L = y**2

dL_dy = 2*y
dy_dz = 1.0
dz_dx = 2.0
dL_dx = dL_dy * dy_dz * dz_dx

print('forward:', z, y, L)
print('dL/dx:', dL_dx)
```

## Single neuron

For `z = wx + b` and `L = (z-y)^2`, compute the gradients manually.

```python
x = 2.0
w = 1.0
b = 0.0
y_true = 7.0

z = w*x + b
loss = (z-y_true)**2

dL_dz = 2*(z-y_true)
dL_dw = dL_dz*x
dL_db = dL_dz

print('prediction:', z)
print('loss:', loss)
print('dL/dw:', dL_dw)
print('dL/db:', dL_db)
```

## Training by gradient descent

Repeat the forward → backward → update loop.

```python
x = 2.0
y_true = 7.0
w = 1.0
b = 0.0
lr = 0.1

for step in range(10):
    z = w*x + b
    loss = (z-y_true)**2
    dL_dz = 2*(z-y_true)
    dL_dw = dL_dz*x
    dL_db = dL_dz
    w -= lr*dL_dw
    b -= lr*dL_db
    print(f'{step:2d}: prediction={z:.4f}, loss={loss:.4f}, w={w:.4f}, b={b:.4f}')
```

## PyTorch automatic differentiation

```python
import torch

x = torch.tensor(2.0)
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
y = torch.tensor(7.0)

z = w*x + b
loss = (z-y)**2
loss.backward()

print('loss:', loss.item())
print('dL/dw:', w.grad.item())
print('dL/db:', b.grad.item())
```
