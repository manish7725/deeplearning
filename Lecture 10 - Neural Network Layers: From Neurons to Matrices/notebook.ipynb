# Lecture 10 — Neural Network Layers: From Neurons to Matrices


```python
import numpy as np

W = np.array([[1., 2.], [3., 4.]])
b = np.array([1., -1.])
x = np.array([2., 1.])
z = W @ x + b
print('W shape:', W.shape)
print('x shape:', x.shape)
print('z:', z)
print('z shape:', z.shape)
```

## See two neurons separately

```python
z1 = W[0] @ x + b[0]
z2 = W[1] @ x + b[1]
print('neuron 1:', z1)
print('neuron 2:', z2)
print('matrix result:', z)
```

## Two linear layers

Notice that two linear transformations can be combined into one.

```python
W1 = np.array([[1., 2.], [0., 1.]])
W2 = np.array([[2., 1.], [1., 3.]])
x = np.array([3., 4.])

two_steps = W2 @ (W1 @ x)
one_step = (W2 @ W1) @ x
print('two layers:', two_steps)
print('combined:', one_step)
```

## PyTorch Linear layer

```python
import torch

layer = torch.nn.Linear(2, 2)
x = torch.tensor([2.0, 1.0])
z = layer(x)
print(z)
print('weight shape:', tuple(layer.weight.shape))
print('bias shape:', tuple(layer.bias.shape))
```
