## Step 1 — Problem
We know the downhill direction from the gradient. We now need a controlled step size and a repeatable update rule.

## Step 2 — Prediction

For `L(w)=(w-3)^2`, start at `w=0`. Predict the first update for learning rates `0.1`, `0.5`, and `1.1`. Which should converge and which should fail?

```python
import numpy as np

def loss(w):
    return (w-3.0)**2

def grad(w):
    return 2*(w-3.0)

w = 0.0
eta = 0.1
w1 = w - eta*grad(w)
print(w1, loss(w), loss(w1))
assert w1 == 0.6
```

## Step 4 — Mathematics

$$w_{new}=w_{old}-\eta\frac{dL}{dw}$$

For `L(w)=(w-3)^2`, $dL/dw=2(w-3)$ and, with $e=w-3$, $e_{new}=(1-2\eta)e$.

```python
# Step 5 — manual updates verified
w = 0.0
for expected in [0.6, 1.08, 1.464]:
    w = w - 0.1*grad(w)
    print(w)
assert np.isclose(w, 1.464)
```

```python
# Step 6 — from-scratch gradient descent
def fit(eta, steps=20):
    w = 0.0
    history = []
    for step in range(steps):
        L = loss(w)
        g = grad(w)
        history.append((step,w,L,g))
        w = w - eta*g
    return np.array(history)

h = fit(0.1, 20)
print(h[:5])
assert h[-1,1] > 2.9 and h[-1,1] < 3.0
```

```python
# Step 7 — visualize the learning trajectory
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))
w_grid=np.linspace(-1,7,300)
plt.plot(w_grid, loss(w_grid), label='loss')
h=fit(0.1, 10)
plt.scatter(h[:,1],h[:,2],s=30,label='updates')
plt.xlabel('w'); plt.ylabel('L(w)'); plt.title('Gradient descent on a quadratic'); plt.legend(); plt.show()
```

```python
# Step 8 — controlled experiment
for eta in [0.1,0.5,1.1]:
    h=fit(eta,12)
    print('eta=',eta,'last w=',h[-1,1],'last loss=',h[-1,2])
```

```python
# Step 9 — change only learning rate and plot loss
plt.figure(figsize=(7,5))
for eta in [0.1,0.5,1.1]:
    h=fit(eta,12)
    plt.plot(range(len(h)),h[:,2],marker='o',label=f'eta={eta}')
plt.xlabel('step'); plt.ylabel('loss'); plt.title('Learning-rate experiment'); plt.legend(); plt.show()
```

## Step 11 — Explain

For the quadratic, the error obeys `e_new=(1-2*eta)e`. Stability therefore requires `|1-2*eta|<1`, giving `0<eta<1`.
