# Lecture 21 — Activation Functions: Bending the Straight Line

This notebook is the hands-on laboratory for the lecture. Predict first, then run the code.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. ReLU

$$f(z)=\max(0,z)$$

```python
z = np.array([-5., -2., -1., 0., 1., 2., 5.])
relu = np.maximum(0, z)
print('z    =', z)
print('ReLU =', relu)
```

## 2. Sigmoid and tanh

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print('sigmoid =', sigmoid(z))
print('tanh    =', np.tanh(z))
```

## 3. Plot the activation functions

```python
x = np.linspace(-6, 6, 400)
plt.figure(figsize=(8, 5))
plt.plot(x, np.maximum(0, x), label='ReLU')
plt.plot(x, sigmoid(x), label='Sigmoid')
plt.plot(x, np.tanh(x), label='Tanh')
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.xlabel('z')
plt.ylabel('f(z)')
plt.title('Activation functions')
plt.legend()
plt.show()
```

## 4. Why linear layers alone do not add expressive power

For scalar layers, compute the composition directly.

```python
# First affine layer: h = 2x + 3
# Second affine layer: y = 4h - 5
# Substitute h into y: y = 4(2x+3)-5 = 8x+7

x = 2
h = 2*x + 3
y = 4*h - 5
y_collapsed = 8*x + 7
print('two affine layers:', y)
print('collapsed affine:', y_collapsed)
```

## 5. A neuron with ReLU

$$a=\operatorname{ReLU}(w^Tx+b)$$

```python
x = np.array([2., -1.])
w = np.array([3., 4.])
b = -2.
z = w @ x + b
a = np.maximum(0, z)
print('pre-activation z =', z)
print('activation a      =', a)
```

## 6. Derivative intuition

For ReLU, the derivative is 0 on the negative side and 1 on the positive side. For sigmoid, the derivative is $\sigma(z)(1-\sigma(z))$.

```python
relu_grad = (x > 0).astype(float)
s = sigmoid(x)
sigmoid_grad = s * (1 - s)
print('x:', x)
print('ReLU derivative:', relu_grad)
print('Sigmoid derivative:', sigmoid_grad)
```
