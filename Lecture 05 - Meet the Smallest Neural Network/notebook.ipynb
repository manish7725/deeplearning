The blog is the textbook; this notebook is the laboratory. Predict first, then run the arithmetic.

```python
import numpy as np

x = np.array([2., 3., 4.])
w = np.array([4., 5., 2.])
b = 1.

z = w @ x + b
print('z =', z)
assert z == 32.
```

## One parameter at a time

Compare how the weight and bias change a one-input neuron.

```python
def predict(x, w, b):
    return w * x + b

x_values = np.array([0.,1.,2.,3.,4.])
print('baseline:', predict(x_values, 3., -2.))
print('change weight:', predict(x_values, 1., -2.))
print('change bias:', predict(x_values, 3., 4.))
```

## One neuron → one layer

Each row of `W` is one neuron's weights.

```python
W = np.array([[1.,2.],[3.,4.],[5.,6.]])
x = np.array([2.,3.])
b = np.array([0.,0.,0.])
z = W @ x + b
print(z)
assert np.allclose(z, [8.,18.,28.])
```

## Batch of examples

This is the matrix form of many neuron evaluations.

```python
X = np.array([[1.,2.],[2.,3.],[3.,4.]])
w = np.array([2.,3.])
b = 1.
y_hat = X @ w + b
print(y_hat)
assert np.allclose(y_hat, [9.,14.,19.])
```

## Parameter counting

```python
def parameter_count(inputs, neurons):
    return inputs * neurons + neurons

print('4 inputs, 3 neurons ->', parameter_count(4,3))
assert parameter_count(4,3) == 15

for d, m in [(1,1),(3,2),(100,10)]:
    print(f'{d} inputs × {m} neurons -> {parameter_count(d,m)} parameters')
```

## Forward pass + loss

Prediction is not learning. Here we make the loss explicit.

```python
x = 3.
y = 7.
w = 0.
b = 0.
y_hat = w * x + b
loss = (y_hat - y) ** 2
print('prediction =', y_hat)
print('loss =', loss)
```

## Mini training experiment

We now use the Chapter 1 gradient-descent idea on a one-neuron model. The point is not the optimizer library—it is seeing the parameters move.

```python
x = np.array([1.,2.,3.,4.])
y = np.array([3.,5.,7.,9.])
w, b = 0., 0.
eta = 0.05

history = []
for step in range(100):
    y_hat = w * x + b
    error = y_hat - y
    loss = np.mean(error**2)
    dw = 2 * np.mean(error * x)
    db = 2 * np.mean(error)
    w -= eta * dw
    b -= eta * db
    history.append(loss)

print('learned w, b =', w, b)
print('final loss =', history[-1])
assert abs(w - 2.) < 0.05
assert abs(b - 1.) < 0.05
```

```python
import matplotlib.pyplot as plt

plt.plot(history)
plt.xlabel('step')
plt.ylabel('MSE')
plt.title('A single neuron learning a line')
plt.show()
```
