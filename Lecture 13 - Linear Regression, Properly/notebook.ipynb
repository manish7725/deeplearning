Fit the house-price line from [`blog.md`](<blog.md>) and compare iterative and closed-form solutions.

## Prediction
For `y = 2x + 1` with small noise, predict a slope near 2 and intercept near 1 before running the notebook.

```python
import numpy as np

x = np.array([1., 2., 3., 4., 5.])
y = np.array([3.2, 4.8, 7.1, 8.9, 11.4])
w, b = 0., 0.
y_hat = w*x + b
print(y_hat)
```

## Mathematics

`J(w,b) = sum((wx + b - y)^2)` and `grad = [2 sum(error*x), 2 sum(error)]`.
The matrix form is `y_hat = X @ beta`. 

```python
x0 = np.array([1., 2., 3., 4.])
y0 = np.array([3., 5., 7., 9.])
w0, b0 = 0., 0.
errors = w0*x0 + b0 - y0
dw = 2*np.sum(errors*x0)
db = 2*np.sum(errors)
assert dw == -140.0
assert db == -48.0
```

```python
w, b = 0., 0.
learning_rate = 0.01
history = []
for step in range(2000):
    pred = w*x + b
    err = pred-y
    loss = np.mean(err**2)
    dw = 2*np.mean(err*x)
    db = 2*np.mean(err)
    w -= learning_rate*dw
    b -= learning_rate*db
    history.append(loss)
print(w, b, history[-1])
assert abs(w-2) < 0.1
assert abs(b-1) < 0.3
```

```python
import matplotlib.pyplot as plt

plt.scatter(x, y, label='data')
plt.plot(x, w*x+b, label='fit')
plt.xlabel('rooms')
plt.ylabel('price (₹ lakh)')
plt.legend()
plt.show()
```

```python
X = np.column_stack([np.ones_like(x), x])
beta_closed = np.linalg.solve(X.T @ X, X.T @ y)
print(beta_closed)
assert np.allclose(beta_closed, np.linalg.lstsq(X, y, rcond=None)[0])
```

```python
y_outlier = y.copy()
y_outlier[-1] = 20.0
beta_outlier = np.linalg.lstsq(X, y_outlier, rcond=None)[0]
print('original:', beta_closed)
print('with outlier:', beta_outlier)
assert beta_outlier[1] > beta_closed[0]*0 + 1.5
```

## Observe
Gradient descent and the closed-form least-squares solution land near the same parameters. The extreme outlier pulls the slope upward.

## Explain
Both methods minimize the same quadratic objective. The closed form solves its stationary equations directly; gradient descent follows the local slope of the same surface.
