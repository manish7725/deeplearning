# Lecture 22 — Why Depth Makes Neural Networks Powerful

Explore composition, hidden representations, and piecewise-linear functions.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Compose two affine functions

$h=2x+3$, then $y=4h-5$.

```python
x = 2.0
h = 2*x + 3
y = 4*h - 5
y_collapsed = 8*x + 7
print(y, y_collapsed)
```

## 2. Add nonlinearity

$h=ReLU(2x+1)$ and $y=3h-4$.

```python
def relu(x):
    return np.maximum(0, x)

xs = np.linspace(-5, 5, 500)
hs = relu(2*xs + 1)
ys = 3*hs - 4

plt.figure(figsize=(8, 5))
plt.plot(xs, ys)
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.title('Two-layer ReLU network')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

## 3. More hidden units create more pieces

```python
def tiny_relu_network(x):
    h1 = relu(x + 2)
    h2 = relu(x)
    h3 = relu(x - 2)
    return 1.5*h1 - 2.0*h2 + 1.0*h3

xs = np.linspace(-5, 5, 500)
ys = tiny_relu_network(xs)
plt.figure(figsize=(8, 5))
plt.plot(xs, ys)
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.title('A piecewise-linear ReLU function')
plt.xlabel('x')
plt.ylabel('network output')
plt.show()
```

## 4. Parameter counting

For a dense layer with `d` inputs and `m` neurons, parameters = `m*d + m`.

```python
layers = [(10, 8), (8, 4), (4, 1)]
counts = [m*d + m for d, m in layers]
print('per-layer:', counts)
print('total:', sum(counts))
```
