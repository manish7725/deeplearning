## 1. What is PyTorch?

PyTorch is a Python library that helps us build and train machine-learning models.

Think of it as a toolbox that gives us:
- **tensors** to store numbers efficiently
- mathematical operations on those numbers
- **automatic differentiation** to calculate gradients
- tools for building neural networks
- support for running computations on GPUs

```python
import torch

print(torch.__version__)
```

## 2. What is a tensor?

A tensor is a container of numbers. You can think of it as a general version of the mathematical objects you already know:

- one number → **scalar**
- a list of numbers → **vector**
- a table of numbers → **matrix**
- more dimensions → higher-dimensional tensor

So a tensor is not a mysterious new mathematical object. It is a convenient way to organize numbers so that PyTorch can perform calculations on them.

```python
scalar = torch.tensor(5)
vector = torch.tensor([10, 20, 30])
matrix = torch.tensor([[1, 2, 3], [4, 5, 6]])

print('scalar:', scalar)
print('vector:', vector)
print('matrix:\n', matrix)
```

## 3. Tensor shape: how is the data arranged?

The **shape** tells us how many numbers exist along each dimension.

For example, a matrix with 2 rows and 3 columns has shape `(2, 3)`.

Shape becomes extremely important in deep learning because neural networks constantly transform tensors from one shape into another.

```python
print('scalar shape:', scalar.shape)
print('vector shape:', vector.shape)
print('matrix shape:', matrix.shape)
```

## 4. Tensor operations are familiar math

We can add, subtract, multiply and take powers of tensors. PyTorch performs these operations element by element when the shapes are compatible.

```python
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([10.0, 20.0, 30.0])

print('a + b =', a + b)
print('a * b =', a * b)
print('a squared =', a ** 2)
print('sum of a =', torch.sum(a))
```

## 5. Matrix multiplication

In neural networks, matrix multiplication is everywhere.

The `@` operator performs matrix multiplication in Python. PyTorch also provides `torch.matmul`.

```python
A = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
B = torch.tensor([[10.0, 20.0], [30.0, 40.0]])

C = A @ B
print(C)
```

## 6. Why does PyTorch need gradients?

Imagine a model has a dial called `w`. We change `w`, the model makes a prediction, and we calculate a loss.

The gradient tells us how the loss changes when we change the dial. That gives the learning algorithm a direction in which to move the dial.

PyTorch can calculate this derivative for us using **automatic differentiation** (autograd).

```python
x = torch.tensor(3.0, requires_grad=True)
y = x ** 2

y.backward()

print('x =', x.item())
print('y = x^2 =', y.item())
print('dy/dx =', x.grad.item())
```

We know mathematically that:

`y = x²`

so:

`dy/dx = 2x`

At `x = 3`, the gradient is `6`. PyTorch found this automatically.

This is one of the key ideas behind modern deep learning: **we can build a complicated calculation and ask PyTorch to calculate the gradients needed to learn its parameters.**

## 7. A tiny learning example

Let's make the idea concrete. Suppose the true relationship is:

`price = 2 × rooms`

Our model starts with the wrong value `w = 0.5`. We will use the loss and gradient to improve it.

```python
rooms = torch.tensor([1.0, 2.0, 3.0, 4.0])
prices = torch.tensor([2.0, 4.0, 6.0, 8.0])

w = torch.tensor(0.5, requires_grad=True)
predictions = w * rooms
loss = torch.mean((predictions - prices) ** 2)

loss.backward()

print('predictions:', predictions)
print('loss:', loss.item())
print('gradient of w:', w.grad.item())
```

### The learning loop in one picture

**Data → Prediction → Loss → Gradient → Update the dial → Repeat**

You will see this loop again and again in the deep-learning lectures.

## 9. What you should remember

- A **tensor is a container of numbers** arranged in one or more dimensions.
- **Shape** tells us how those numbers are arranged.
- PyTorch makes mathematical operations on tensors easy and efficient.
- `requires_grad=True` asks PyTorch to track calculations for differentiation.
- `.backward()` calculates gradients for the tracked computation.
- Gradients tell learning algorithms how changing model parameters changes the loss.
