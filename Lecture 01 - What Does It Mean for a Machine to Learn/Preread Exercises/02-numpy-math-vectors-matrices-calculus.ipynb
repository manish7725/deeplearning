# Pre-read 2 — NumPy + the Mathematics of ML

## Why are we learning NumPy, vectors, matrices, and calculus?

In the previous notebook we wrote a tiny ML program in plain Python. It worked, but real ML quickly becomes a **numbers problem**: thousands of examples, many features, model parameters, predictions, errors, and gradients.

We need a better way to organize and calculate with those numbers. **NumPy gives us arrays and fast mathematical operations.** Vectors and matrices give us the language used to describe those calculations. Calculus gives us the idea of a **slope**—the information we need to improve model parameters.

Our path is:

**many numbers → arrays → vectors → matrices → model calculation → loss → slope**

You do not need to memorize mathematics here. The goal is to connect each mathematical idea to the ML problem it helps us solve.

## 1. NumPy arrays — move beyond Python lists

**Why:** ML datasets contain many numbers, and we often want to perform the same calculation on all of them. NumPy arrays are designed for this kind of numerical work.

```python
import numpy as np

rooms = np.array([1, 2, 3, 4])
prices = np.array([3, 5, 7, 9])

print(rooms)
print("shape:", rooms.shape)
```

The **shape** tells us how data is arranged. Here we have 4 examples. Later you will see shapes such as `(1000, 5)` meaning 1000 examples and 5 features per example.

## 2. Vector arithmetic — calculate many predictions at once

**Why:** Instead of writing a loop for every number, we want one mathematical operation to describe the whole calculation.

For example, multiplying every room count by 2 is simply `2 × rooms`.

```python
rooms = np.array([1, 2, 3, 4])

predictions = 2 * rooms + 1
print(predictions)
```

This is already closer to how ML is written: **one expression describes a calculation over many examples.**

## 3. Dot product — combine features and weights

**Why:** A real house may have several features: rooms, bathrooms, area, age, and so on. A model gives each feature a weight and combines them into a prediction.

The dot product is the small mathematical operation that performs this combination.

```python
features = np.array([3, 2, 100])
weights = np.array([2, 5, 0.1])

score = features @ weights
print(score)
```

Read `features @ weights` as:

`3×2 + 2×5 + 100×0.1`

**Why this matters:** Neural networks perform huge numbers of similar weighted combinations. The dot product is one of their basic building blocks.

## 4. Matrices — represent a whole dataset

**Why:** One house is one feature vector. Thousands of houses form a matrix: **rows = examples, columns = features**. This lets us express a large ML calculation compactly.

```python
X = np.array([
    [2, 1],
    [3, 2],
    [4, 2]
])

print(X)
print("shape:", X.shape)
```

Here the 3 rows are houses and the 2 columns could be `rooms` and `bathrooms`.

Now give the two features weights and calculate all three scores at once.

```python
weights = np.array([2, 5])

scores = X @ weights
print(scores)
```

**Why:** This is the bridge from simple Python to ML mathematics: **data matrix × parameter vector → predictions**.

## 5. Loss — turn many errors into one number

**Why:** A model needs a score that tells us how good or bad its predictions are. That score is called **loss**. A common choice is mean squared error (MSE).

```python
true = np.array([3, 5, 7, 9])
predicted = np.array([2, 6, 7, 8])

errors = predicted - true
mse = np.mean(errors ** 2)

print("errors:", errors)
print("MSE:", mse)
```

Notice the progression:

**prediction → error → squared error → average → one loss number**

Now we have a number we can try to make smaller.

## 6. Calculus — how do we know which way to improve?

**Why:** Suppose changing a model parameter changes the loss. Should we increase the parameter or decrease it? We need a direction.

Calculus gives us that direction through the idea of a **slope**.

Start with the simple question:

**If I move the input a tiny amount, how much does the output change?**

That is approximately `change in output ÷ change in input`. 

```python
def f(x):
    return x ** 2

x = 3
h = 0.001
slope = (f(x + h) - f(x)) / h

print("approximate slope:", slope)
```

For `f(x) = x²`, the exact derivative is `2x`, so at `x = 3` the slope is 6. The small-step experiment gets us close to that answer.

## 7. Partial derivatives — one model dial at a time

**Why:** ML models usually have many parameters. We can ask how the loss changes when we move **one parameter while keeping the others fixed**.

That is the intuition behind a partial derivative.

```python
def loss(w, b):
    return (w - 2) ** 2 + (b - 1) ** 2

w, b = 0.5, 0.0
h = 0.0001

dw = (loss(w + h, b) - loss(w, b)) / h
db = (loss(w, b + h) - loss(w, b)) / h

print("change in loss when w moves ≈", dw)
print("change in loss when b moves ≈", db)
```

The two slopes give us two pieces of information about which way the loss changes. With many parameters, we collect these slopes into a **gradient**.

That leads directly to the next idea: **use the gradient to update the model parameters.**
