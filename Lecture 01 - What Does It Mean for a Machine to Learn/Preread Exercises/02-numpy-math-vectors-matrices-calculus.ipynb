## 2. Vector arithmetic

If we have two vectors, NumPy lets us add them element by element.

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print("a + b =", a + b)
print("a - b =", a - b)
print("2a =", 2 * a)
```

### One very useful idea

When you write `2 * a`, NumPy multiplies **every element** by 2. This is the same idea we use when a model multiplies every input by a weight.

## 3. Dot product

The dot product combines two vectors into one number.

For `[1, 2, 3]` and `[4, 5, 6]`:

`1×4 + 2×5 + 3×6 = 32`

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.dot(a, b))
print(a @ b)  # another NumPy way to write a dot product
```

## 4. Matrices

A matrix is a rectangle of numbers. In NumPy, a matrix is usually represented by a 2-D array.

```python
A = np.array([[1, 2],
              [3, 4]])

print(A)
print("shape:", A.shape)
print("first row:", A[0])
print("bottom-right value:", A[1, 1])
```

### Matrix addition
Two matrices with the same shape can be added element by element.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[10, 20], [30, 40]])

print(A + B)
```

## 5. Matrix multiplication

Matrix multiplication is different from element-by-element multiplication. The `@` operator means matrix multiplication in NumPy.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(A @ B)
```

For the top-left value, NumPy calculates:

`1×5 + 2×7 = 19`

That is a dot product between a row of `A` and a column of `B`. This small idea becomes extremely important in neural networks.

## 6. Solve a small system of equations

Suppose:

`2x + y = 5`
`x + y = 3`

We can write this as `A @ [x, y] = b`. NumPy can solve it for us.

```python
A = np.array([[2, 1],
              [1, 1]])
b = np.array([5, 3])

solution = np.linalg.solve(A, b)
print(solution)
```

### Check your answer
Put the solution back into the original equations. Does it really produce 5 and 3?

```python
x, y = solution
print(2*x + y)
print(x + y)
```

## 7. Mean, sum, and square

These are small operations you will use constantly when calculating loss.

```python
errors = np.array([-3, -5, -7, -9])

print("squared:", errors ** 2)
print("sum of squared errors:", np.sum(errors ** 2))
print("mean squared error:", np.mean(errors ** 2))
```

## 8. Calculus without fear: slope as a small experiment

Before learning formal derivatives, think about slope as:

**change in output ÷ change in input**

For `f(x) = x²`, take a small step `h` and compare `f(x+h)` with `f(x)`.

```python
def f(x):
    return x**2

x = 3
h = 0.001

approx_slope = (f(x + h) - f(x)) / h
print(approx_slope)
```

At `x = 3`, the exact derivative of `x²` is `2x = 6`. Our tiny-step experiment should be very close to 6.

```python
for h in [0.1, 0.01, 0.001, 0.0001]:
    slope = (f(3 + h) - f(3)) / h
    print("h =", h, "slope ≈", slope)
```

## 9. Partial derivatives: one dial at a time

Imagine a function with two inputs:

`L(w, b) = (w - 2)² + (b - 1)²`

A partial derivative asks: **what happens if I move one input while keeping the other fixed?**

We can estimate each slope with a tiny step.

```python
def L(w, b):
    return (w - 2)**2 + (b - 1)**2

w, b = 0.5, 0.0
h = 0.0001

dw = (L(w + h, b) - L(w, b)) / h
db = (L(w, b + h) - L(w, b)) / h

print("slope with respect to w ≈", dw)
print("slope with respect to b ≈", db)
```
