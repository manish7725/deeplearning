## 5. Functions — package an idea

A function is a reusable piece of code. This will become very important when we build models.

```python
def predict_price(rooms, w, b):
    return w * rooms + b

print(predict_price(3, 2, 1))
print(predict_price(5, 2, 1))
```

## 6. A tiny machine-learning calculation

Now combine variables, lists, loops, and a function. We will calculate the predictions of a simple model for four houses.

```python
rooms = [1, 2, 3, 4]
true_prices = [3, 5, 7, 9]

w = 1
b = 0

for x, y in zip(rooms, true_prices):
    y_hat = predict_price(x, w, b)
    error = y_hat - y
    print("x =", x, "prediction =", y_hat, "truth =", y, "error =", error)
```
