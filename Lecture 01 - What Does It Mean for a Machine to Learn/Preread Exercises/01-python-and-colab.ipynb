A machine-learning idea starts as a simple thought: **take some data, calculate a prediction, measure how wrong it is, and improve the model.**

We need a way to tell the computer exactly how to perform those steps. Python is the language we will use throughout this course to express those ideas as executable code.

Our path is:

**Python → data → calculation → prediction → error → machine-learning model**

## 2. Variables — give data a name

**Why:** ML models work with many values: number of rooms, age, price, weight, probability, and so on. Variables let us name those values so we can use them in calculations.

```python
rooms = 3
price = 7

print(rooms)
print(price)
```

Now express a tiny prediction rule:

`prediction = price_per_room × rooms + base_price`

**Why:** This is the first glimpse of a model: a rule that turns input data into a prediction.

```python
rooms = 5
price_per_room = 2
base_price = 1

prediction = price_per_room * rooms + base_price
print(prediction)
```

## 3. Lists — work with many examples

**Why:** A real ML model does not learn from one house. It learns from many examples. A Python list lets us keep several values together.

```python
rooms = [1, 2, 3, 4]
prices = [3, 5, 7, 9]

print(rooms)
print(prices)
```

Each position connects an input with its target:

`1 room → 3`
`2 rooms → 5`
`3 rooms → 7`
`4 rooms → 9`

**Why:** This input/target relationship is the basic shape of supervised learning data.

## 4. Loops — repeat the same calculation

**Why:** Once we have many examples, we need the computer to perform the same operation for every example. A loop does that.

```python
for x in rooms:
    print("House has", x, "rooms")
```

Now calculate a prediction for every house using `prediction = 2 × rooms + 1`.

**Why:** Later, training will repeatedly calculate predictions for batches of data.

```python
for x in rooms:
    prediction = 2 * x + 1
    print(prediction)
```

## 5. Functions — turn a model idea into reusable code

**Why:** We do not want to rewrite the prediction formula every time. A function packages the rule so we can call it with different inputs and parameters.

```python
def predict_price(rooms, w, b):
    return w * rooms + b

print(predict_price(3, 2, 1))
print(predict_price(5, 2, 1))
```

Notice the important idea: `w` and `b` are **dials**. Change them and the prediction changes. In machine learning, these are model parameters that training will learn.

## 6. Prediction → error

**Why:** A model becomes learnable only when we can tell it how well it is doing. Compare a prediction with the true answer.

```python
rooms = [1, 2, 3, 4]
true_prices = [3, 5, 7, 9]

w = 1
b = 0

for x, y in zip(rooms, true_prices):
    prediction = predict_price(x, w, b)
    error = prediction - y
    print("input =", x, "prediction =", prediction, "truth =", y, "error =", error)
```

We have now written the beginning of an ML program:

**data → model → prediction → error**
