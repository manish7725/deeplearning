# Pre-read 3 — Visualization with Matplotlib

## Why are we learning visualization before machine learning?

By now we can calculate predictions and loss. But numbers alone can hide patterns. A graph can make a pattern, a bad model, an error, or a valley in the loss immediately visible.

Visualization is therefore not decoration. It is a **thinking and debugging tool** for ML. We will use it to answer questions such as:
- What pattern exists in the data?
- Does my model fit the data?
- Where are the prediction errors?
- Which parameter value gives lower loss?
- What does the learning process look like?

Our path is:

**numbers → picture → pattern → model intuition → better decisions**

## 1. Plot the data

**Why:** Before asking a model to learn from data, we should look at the data. A scatter plot lets us see whether inputs and targets appear related.

```python
import numpy as np
import matplotlib.pyplot as plt

rooms = np.array([1, 2, 3, 4])
prices = np.array([3, 5, 7, 9])

plt.scatter(rooms, prices)
plt.xlabel("Rooms")
plt.ylabel("Price")
plt.title("House data")
plt.show()
```

**Read the picture:** each dot is one training example. The visual question is: **can you see a relationship between rooms and price?**

## 2. Put the model on the same graph

**Why:** A model is supposed to explain or predict the data. Plotting the model and the data together lets us inspect whether its predictions make sense.

Our model is `prediction = w*x + b`. Start with `w = 1, b = 0`. 

```python
w = 1
b = 0
x_line = np.linspace(0, 5, 100)
y_line = w * x_line + b

plt.scatter(rooms, prices, label="Data")
plt.plot(x_line, y_line, label="Model")
plt.xlabel("Rooms")
plt.ylabel("Price")
plt.legend()
plt.show()
```

### Practice
Change `w` to 2 and `b` to 1.

**Why:** You are learning to see how model parameters move the prediction line. This will matter when training starts changing those parameters automatically.

```python
w = 2
b = 1
y_line = w * x_line + b

plt.scatter(rooms, prices, label="Data")
plt.plot(x_line, y_line, label="Model")
plt.xlabel("Rooms")
plt.ylabel("Price")
plt.legend()
plt.show()
```

## 3. See prediction errors

**Why:** The distance between a prediction and the true value is an error. Seeing those gaps makes the idea of loss concrete.

```python
w = 1
b = 0
predictions = w * rooms + b

plt.scatter(rooms, prices, label="True")
plt.scatter(rooms, predictions, label="Prediction")

for x, true, pred in zip(rooms, prices, predictions):
    plt.plot([x, x], [true, pred], linestyle="--")

plt.xlabel("Rooms")
plt.ylabel("Price")
plt.title("Prediction errors")
plt.legend()
plt.show()
```

Try `w = 2, b = 1`.

**Question:** What happens to the error gaps?

This is the visual version of **prediction → error → loss**.

```python
w = 2
b = 1
predictions = w * rooms + b

for x, true, pred in zip(rooms, prices, predictions):
    plt.plot([x, x], [true, pred], linestyle="--")

plt.scatter(rooms, prices, label="True")
plt.scatter(rooms, predictions, label="Prediction")
plt.xlabel("Rooms")
plt.ylabel("Price")
plt.title("Errors after changing the model")
plt.legend()
plt.show()
```

## 4. Plot a function and see a valley

**Why:** We want the model to reduce loss. A graph of loss helps us visualize the goal: move toward the bottom of a valley.

First, look at the simple function `y = x²`. 

```python
x = np.linspace(-3, 3, 200)
y = x ** 2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("A simple valley: y = x²")
plt.show()
```

The lowest point is the place where the value is smallest. In ML, we create a similar picture for **loss versus a model parameter**.

## 5. Visualize loss as a model parameter changes

**Why:** A model parameter is a dial. If we try different settings for the dial, the loss changes. Plotting those values lets us see which settings are better.

```python
w_values = np.linspace(0, 4, 201)
loss_values = 7.5 * (w_values - 2) ** 2

plt.plot(w_values, loss_values)
plt.xlabel("w")
plt.ylabel("Loss")
plt.title("Loss as the model dial w changes")
plt.show()
```

### Practice
Use NumPy to find the `w` value that gives the smallest loss in this grid.

**Why:** You are connecting a visual question—“where is the bottom?”—to a numerical question—“which value has the smallest loss?”

```python
best_index = np.argmin(loss_values)
print("best w ≈", w_values[best_index])
print("smallest loss ≈", loss_values[best_index])
```

## 6. Watch the model move toward a better setting

**Why:** Learning is not just about finding a good answer. It is about repeatedly changing parameters to reduce loss. We can make that process visible by trying several values of `w`.

```python
for w in [0, 1, 1.5, 2, 2.5, 3]:
    predictions = w * rooms + 1
    loss = np.mean((predictions - prices) ** 2)
    print(f"w={w}, loss={loss:.3f}")

    plt.scatter(rooms, prices)
    plt.plot(x_line, w * x_line + 1)
    plt.xlabel("Rooms")
    plt.ylabel("Price")
    plt.title(f"w={w}, loss={loss:.3f}")
    plt.show()
```

The important pattern is not the plotting code. It is this:

**change parameter → make predictions → calculate loss → compare → choose a better direction**
