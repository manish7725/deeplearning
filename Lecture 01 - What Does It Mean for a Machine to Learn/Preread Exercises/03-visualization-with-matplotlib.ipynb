# Preread 3 — Visualization with Matplotlib

A table of numbers can tell us **what** happened. A graph can often show us **what is happening** immediately.

In Lecture 01 we will talk about predictions, errors, loss, valleys, and slopes. This notebook gives you the basic visualization skills needed to see those ideas.

> 🎯 **Goal:** Learn to plot points, draw a line, plot a function, label a graph, and visualize a loss curve.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Your first plot

Let's put our four houses on a graph.

```python
rooms = np.array([1, 2, 3, 4])
prices = np.array([3, 5, 7, 9])

plt.scatter(rooms, prices)
plt.xlabel("Rooms")
plt.ylabel("Price (₹ lakh)")
plt.title("House prices")
plt.show()
```

### What are you looking at?

- Each **dot** is one house.
- The horizontal axis is the number of rooms.
- The vertical axis is the price.

The visual question is: **do the dots follow a pattern?**

## 2. Draw the model's line

Our simple model is `ŷ = wx + b`. Start with `w = 1` and `b = 0`.

```python
w = 1
b = 0

x_line = np.linspace(0, 5, 100)
y_line = w * x_line + b

plt.scatter(rooms, prices, label="True prices")
plt.plot(x_line, y_line, label="Model")
plt.xlabel("Rooms")
plt.ylabel("Price (₹ lakh)")
plt.title("Data and model")
plt.legend()
plt.show()
```

## 3. Visualize a mathematical function

A function can also be drawn as a curve. Let's look at `y = x²`.

```python
x = np.linspace(-3, 3, 200)
y = x**2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("y = x²")
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.show()
```

Notice the **valley**. The lowest point is at `x = 0`. This is the visual idea we will later use for loss landscapes.

## 4. Visualize loss as the model changes

For Lecture 01, hold `b = 1`. Our four houses produce the loss:

`L(w) = 7.5 × (w - 2)²`

Let's draw it.

```python
w_values = np.linspace(0, 4, 200)
loss_values = 7.5 * (w_values - 2)**2

plt.plot(w_values, loss_values)
plt.xlabel("w")
plt.ylabel("Loss")
plt.title("Loss as w changes")
plt.show()
```

### Find the bottom

Use NumPy to find which value of `w` in our grid gives the smallest loss.

```python
best_index = np.argmin(loss_values)
best_w = w_values[best_index]
best_loss = loss_values[best_index]

print("best w ≈", best_w)
print("smallest loss ≈", best_loss)
```

Because we used a grid of many values, the answer is approximately 2. If we include `w = 2` exactly, the loss is exactly zero.

```python
w_values = np.linspace(0, 4, 201)  # includes 2 exactly
loss_values = 7.5 * (w_values - 2)**2

index = np.argmin(loss_values)
print("w =", w_values[index])
print("loss =", loss_values[index])
```

## 5. See prediction errors visually

A vertical gap between a prediction and the true value is an error. Let's draw those gaps.

```python
w = 1
b = 0
predictions = w * rooms + b

plt.scatter(rooms, prices, label="True")
plt.scatter(rooms, predictions, label="Prediction")

for x, true, pred in zip(rooms, prices, predictions):
    plt.plot([x, x], [true, pred], linestyle="--")

plt.xlabel("Rooms")
plt.ylabel("Price (₹ lakh)")
plt.title("Prediction errors")
plt.legend()
plt.show()
```

## 6. A small animation-like experiment

We can repeatedly redraw the model for different values of `w`. Even without a real animation, watching the line move helps build intuition.

```python
for w in [0, 1, 1.5, 2, 2.5, 3]:
    y_line = w * x_line + 1
    loss = np.mean((w * rooms + 1 - prices)**2)
    print(f"w={w:>3}, loss={loss:.3f}")

    plt.scatter(rooms, prices)
    plt.plot(x_line, y_line)
    plt.xlabel("Rooms")
    plt.ylabel("Price (₹ lakh)")
    plt.title(f"w={w}, loss={loss:.3f}")
    plt.show()
```
