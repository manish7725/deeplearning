```python
import numpy as np
import matplotlib.pyplot as plt

e1 = np.array([1., 0.])
e2 = np.array([0., 1.])
x = np.array([3., 4.])

A = np.array([[2., 0.], [0., 2.]])
print('A x =', A @ x)
assert np.allclose(A @ x, [6., 8.])
```

## Basis-vector experiment

Predict where the two basis vectors go, then run the cell.

```python
matrices = {
    'identity': np.array([[1.,0.],[0.,1.]]),
    'scale': np.array([[2.,0.],[0.,3.]]),
    'reflect': np.array([[0.,1.],[1.,0.]]),
    'rotate90': np.array([[0.,-1.],[1.,0.]]),
    'shear': np.array([[1.,1.],[0.,1.]])
}
for name, M in matrices.items():
    print(name, 'e1 ->', M @ e1, 'e2 ->', M @ e2, 'det =', np.linalg.det(M))
```

## Visualize a transformed grid

```python
def grid_points(n=11):
    t = np.linspace(-2, 2, n)
    pts = []
    for a in t:
        pts.append(np.column_stack([np.full_like(t, a), t]))
    for b in t:
        pts.append(np.column_stack([t, np.full_like(t, b)]))
    return np.vstack(pts)

P = grid_points()
M = matrices['shear']
Q = P @ M.T

fig, ax = plt.subplots(figsize=(6,6))
ax.scatter(P[:,0], P[:,1], s=8, alpha=.25)
ax.scatter(Q[:,0], Q[:,1], s=8, alpha=.55)
ax.axhline(0, linewidth=0.8)
ax.axvline(0, linewidth=0.8)
ax.set_aspect('equal')
ax.set_title('Original points and transformed points')
plt.show()
```

## Determinant and area

A unit square is transformed by each matrix. Predict the area scale before running.

```python
for name, M in matrices.items():
    print(f'{name:>9}: determinant={np.linalg.det(M): .2f}, area scale={abs(np.linalg.det(M)): .2f}')

collapse = np.array([[1.,2.],[2.,4.]])
print('collapse det =', np.linalg.det(collapse))
assert abs(np.linalg.det(collapse)) < 1e-10
```

## Composition: rightmost happens first

```python
A = matrices['rotate90']
B = matrices['scale']
v = np.array([1., 0.])

BA_v = B @ (A @ v)
AB_v = A @ (B @ v)
print('B(A v) =', BA_v)
print('A(B v) =', AB_v)
assert not np.allclose(BA_v, AB_v)
```

## Two affine layers without activation

Verify numerically that two affine transformations can be combined into one affine transformation.

```python
W1 = np.array([[1.,2.],[0.,1.]])
b1 = np.array([1.,-1.])
W2 = np.array([[2.,0.],[1.,3.]])
b2 = np.array([2.,4.])
x = np.array([3.,5.])

y_two = W2 @ (W1 @ x + b1) + b2
W = W2 @ W1
b = W2 @ b1 + b2
y_one = W @ x + b
print('two layers:', y_two)
print('one affine map:', y_one)
assert np.allclose(y_two, y_one)
```
