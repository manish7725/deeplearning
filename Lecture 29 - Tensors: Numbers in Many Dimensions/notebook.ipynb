## Step 1 — Problem

## Step 2 — Prediction

Before running code, predict the shape, rank and number of values for `(4, 3, 8, 8)`. Predict the result of adding a `(3,)` bias to a `(2, 3)` matrix.

```python
import numpy as np

X = np.arange(4 * 3 * 8 * 8).reshape(4, 3, 8, 8)
print('shape:', X.shape)
print('ndim:', X.ndim)
print('numel:', X.size)
assert X.shape == (4, 3, 8, 8)
assert X.ndim == 4
assert X.size == 4 * 3 * 8 * 8
```

## Step 4 — Mathematics

For shape $(d_1,\ldots,d_k)$, the number of values is $\prod_i d_i$. Broadcasting aligns compatible trailing axes.

```python
M = np.array([[1., 2., 3.], [4., 5., 6.]])
b = np.array([10., 20., 30.])

loop_result = np.empty_like(M)
for r in range(M.shape[0]):
    for c in range(M.shape[1]):
        loop_result[r, c] = M[r, c] + b[c]

broadcast_result = M + b
assert np.allclose(loop_result, broadcast_result)
assert np.allclose(broadcast_result, np.array([[11.,22.,33.],[14.,25.,36.]]))
print(broadcast_result)
```

```python
# Step 6 — first implementation: explicit shape reasoning
shape = (4, 3, 8, 8)
total = 1
for dimension in shape:
    total *= dimension
assert total == np.prod(shape)
assert total == 4 * 3 * 8 * 8
print('values:', total)
```

```python
# Step 7 — visualization: one tiny image tensor
tiny = np.arange(2 * 3 * 2 * 2).reshape(2, 3, 2, 2)
print('first image, first channel:')
print(tiny[0, 0])
```

```python
# Step 8 — experiment: change only batch size
X8 = np.zeros((8, 3, 8, 8))
print('original:', X.shape)
print('changed batch:', X8.shape)
```

```python
# Step 9 — verify exactly one axis moved
assert X.shape[1:] == X8.shape[1:]
assert X.shape[0] == 4
assert X8.shape[0] == 8
print('only the batch axis changed')
```

## Step 10 — Observe

The first three axes stayed fixed when only the batch size changed. Broadcasting changed the result without changing the mathematical meaning of the bias.

## Step 11 — Explain

A tensor is a multidimensional numerical container; shape tells us how axes are arranged. Broadcasting lets a compatible smaller shape act repeatedly along shared axes.
