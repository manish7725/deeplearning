```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.datasets import make_circles

X = np.array([[1,2],[2,1],[4,4],[5,5],[1,1],[5,4]], dtype=float)
y = np.array([-1,-1,1,1,-1,1])
model = make_pipeline(StandardScaler(), SVC(kernel='linear', C=1.0))
model.fit(X, y)
print('predictions:', model.predict(X))
```

## 4. Mathematics

Decision boundary: $w^T x+b=0$. Full margin: $2/||w||$. Kernel: $K(x,z)=phi(x)^T phi(z)$.


```python
w_norm = 5.0
margin = 2.0 / w_norm
assert np.isclose(margin, 0.4)
print('full margin when ||w||=5:', margin)

x = np.array([1.,2.])
z = np.array([3.,4.])
kernel_value = (x @ z) ** 2
assert np.isclose(kernel_value, 121.0)
print('polynomial kernel value:', kernel_value)
```

```python
# Explicitly compute the quadratic feature map used in the blog.
def phi(v):
    x1, x2 = v
    return np.array([x1**2, np.sqrt(2)*x1*x2, x2**2])

explicit = phi(x) @ phi(z)
implicit = (x @ z) ** 2
assert np.isclose(explicit, implicit)
print('explicit feature dot product:', explicit)
print('kernel shortcut:', implicit)
```

```python
# Visualization: linear SVM margin and support vectors.
svm = SVC(kernel='linear', C=1.0).fit(X, y)
xx = np.linspace(0, 6, 200)
yy = np.linspace(0, 6, 200)
XX, YY = np.meshgrid(xx, yy)
grid = np.c_[XX.ravel(), YY.ravel()]
Z = svm.decision_function(grid).reshape(XX.shape)

plt.figure()
plt.scatter(X[:,0], X[:,1], c=y)
plt.contour(XX, YY, Z, levels=[-1,0,1])
plt.scatter(svm.support_vectors_[:,0], svm.support_vectors_[:,1], s=100, facecolors='none', edgecolors='black')
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Linear SVM: boundary, margins, support vectors')
plt.show()
```

```python
# Experiment: change exactly one variable, C.
C_values = [0.01, 0.1, 1, 10, 100]
support_counts = []
for C in C_values:
    m = SVC(kernel='linear', C=C).fit(X, y)
    support_counts.append(len(m.support_))
print(dict(zip(C_values, support_counts)))
```

```python
# Nonlinear data: concentric circles.
X_circle, y_circle = make_circles(n_samples=300, factor=0.35, noise=0.08, random_state=7)
linear = SVC(kernel='linear', C=1.0).fit(X_circle, y_circle)
rbf = SVC(kernel='rbf', C=1.0, gamma='scale').fit(X_circle, y_circle)

print('linear training accuracy:', linear.score(X_circle, y_circle))
print('RBF training accuracy:', rbf.score(X_circle, y_circle))
```
